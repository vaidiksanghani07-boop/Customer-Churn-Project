```python
import pandas as pd 
import numpy as np 
from sklearn.model_selection import train_test_split 
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix ,recall_score , precision_score , f1_score , accuracy_score, classification_report


data  = pd.read_csv("../data/Telco-Customer-Churn.csv")
data["TotalCharges"] = pd.to_numeric(data["TotalCharges"],errors="coerce")
data["TotalCharges"] = data["TotalCharges"].fillna(0)

data = data.drop("customerID", axis=1)

data = pd.get_dummies(data, drop_first=True)


X = data.drop("Churn_Yes", axis=1)
y = data["Churn_Yes"]

X_train,X_test,y_train,y_test = train_test_split(X , y , test_size = 0.2 , random_state = 1 , stratify=y)

model = RandomForestClassifier(random_state = 1)
model.fit(X_train , y_train)
y_pred_r = model.predict(X_test)
print(y_pred_r[:10])

print("accuracy_score :" ,accuracy_score(y_test , y_pred_r))
print("precision :",precision_score(y_test , y_pred_r))
print("recall :",recall_score(y_test , y_pred_r))
print("f1 score : ",f1_score(y_test , y_pred_r))
print("\nclassification report")
print(classification_report(y_test , y_pred_r))
print("\n confusion matrix :")
print(confusion_matrix(y_test , y_pred_r))
```

```python
from sklearn.model_selection import GridSearchCV


params = {
    "n_estimators": [50, 100, 200],
    "max_depth": [2, 5, 10],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [2, 5, 10]
}

grid_search = GridSearchCV(RandomForestClassifier(criterion = "gini" , random_state = 1 , class_weight = "balanced"), param_grid = params , n_jobs = -1 , cv = 5 , scoring = "f1")
grid_search.fit(X_train , y_train)
print(grid_search.best_params_)
print(grid_search.best_score_)

```

```python
best_model = grid_search.best_estimator_
y_pred_grid = best_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_grid))
print("Precision:", precision_score(y_test, y_pred_grid))
print("Recall:", recall_score(y_test, y_pred_grid))
print("F1 Score:", f1_score(y_test, y_pred_grid))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_grid))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_grid))
```

```python
import joblib

joblib.dump(best_model , "../models/random_forest.pkl")
```
