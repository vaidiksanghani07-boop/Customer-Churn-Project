```python
import pandas as pd 
import numpy as np 
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix ,recall_score , precision_score , f1_score , accuracy_score, classification_report


data  = pd.read_csv("../data/Telco-Customer-Churn.csv")
data["TotalCharges"] = pd.to_numeric(data["TotalCharges"],errors="coerce")

data["TotalCharges"] = data["TotalCharges"].fillna(0)
customer_ids = data["customerID"]

data = data.drop("customerID", axis=1)

data = pd.get_dummies(data, drop_first=True)


X = data.drop("Churn_Yes", axis=1)
y = data["Churn_Yes"]

X_train,X_test,y_train,y_test = train_test_split(X , y , test_size = 0.2 , random_state = 1 , stratify=y)

d_model = DecisionTreeClassifier(criterion="gini" , max_depth = 5 , min_samples_leaf = 5,min_samples_split = 10 ,random_state =1)

d_model.fit(X_train , y_train)
y_pred_d = d_model.predict(X_test)

print("accuracy" , accuracy_score(y_test , y_pred_d))
print("recall", recall_score(y_test , y_pred_d))
print("f1 score" , f1_score(y_test , y_pred_d))
print("precision" , precision_score(y_test , y_pred_d))
print("\nclassification report:")
print(classification_report(y_test , y_pred_d))
print("confusion matrix:")
print(confusion_matrix(y_test , y_pred_d))
 


```

```python
from sklearn.model_selection import GridSearchCV

params = {
    "max_depth": [3,5,7,None],
    "min_samples_split" :[2 , 5, 10],
    "min_samples_leaf" :[1,2,5]
}

grid_search = GridSearchCV(DecisionTreeClassifier(criterion = "gini" , random_state = 1),param_grid = params,cv = 5 ,n_jobs = 1,scoring="accuracy")
grid_search.fit(X_train,y_train)
print(grid_search.best_params_)
print(grid_search.best_score_)

best_d_model = grid_search.best_estimator_

y_pred = best_d_model.predict(X_test)

print("accuracy :" , accuracy_score(y_test , y_pred))
print("precision :",precision_score(y_test , y_pred))
print("recall :" ,recall_score(y_test , y_pred))
print("f1 score :",f1_score(y_test , y_pred))
print("\n classification report:")
print(classification_report(y_test , y_pred))
print("\nconfusion matix")
print(confusion_matrix(y_test , y_pred))
```
